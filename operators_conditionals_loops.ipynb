##ASSIGNAMENT OPERATORS
these are unary operators performed on one operand to change value of variable includes:+=,-=,*=,/=,%=,//=

```python
a=int(input("enter a value"))
a+=10
print(a)
a-=20
print(a)
a*=10
print(a)
a/=10
print(a)
b=int(input("enter b value"))
b//=35
print(b)
b%=10
print(b)
```

##RELATIONAL OPERATORS
preformed on two operands
includes:<,>,<=,>=,==,!=

```python
a=5
b=14
print(a<b)
print(a>b)
print(a<=b)
c=10
d=10
print(c>=d)
print(c==d)
print(c!=d)
```

##program on match case

---



```python
while(1):
  day=int(input("enter any number"))
  match day:
    case 1:
      print("monday")
    case 2:
      print("tuesday")
    case 3:
      print("wednesday")
    case 4:
      print("thursday")
    case 5:
      print("friday")
    case 6:
      print("saturday")
    case 7:
      print("exit")
      break
    case _:
      print("default")

```
