1.WRITING IN A FILE

```python
file=open("data.txt","w")
content="hi chandu"
file.write(content)
file.close()
```

2.READING A FILE

```python
with open ("data.txt","r") as file:
  content1=file.read(5)
  print(content1)
  content=file.read()
  print(content)
```

3.APPENDING A FILE

```python
file=open("data.txt","a")
file.write(" hello chandu\nhi vihaan")
file.close()
```

4.FOR CHECKING BUILTIN METHODS

```python
file.closed #to know whether file is closed or not
file.mode #to know mode
file.name #to know name
```

5.WRITE AND WRITE LINES

```python
file=open("data.txt","w")
file.write("8008324729")
#after this overwites and prints in new line below content
file.writelines(["\nhi vishvika","\nhi vlvvnsv","\nwe are all family"])
file.close()
```

6.READ AND READLINES

```python
with open ("data.txt","r") as file:
  content=file.read(5)
  print(content)
  content1=file.read()
  print(content1)
  print(file.tell())
  file.seek(0)
  print(file.tell())
  content2=file.readlines()
  print(content2)
```
