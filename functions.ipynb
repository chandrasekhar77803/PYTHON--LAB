#write a program to check given number is palindrome or not

```python
def check(rev,n):
  m=n
  while(n!=0):
    rev=rev*10+(n%10)
    n=n//10
  if(m==rev):
    return 1
  else:
    return 0


n=int(input())
k=check(0,n)
if(k==1):
  print("palindrome")
else:
  print("not a palindrome")

```

PROGRAM TO REVERSE A NUMBER




```python
def rev(n):
  rev=0
  while(n>0):
    r=n%10
    rev=rev*10+r
    n=n//10
  return rev
n=int(input("enter a number"))
rev(n)
```

#program that reverse a string

```python
def rev_str(s):
  rev=s[::-1]
  return rev
s=input()
rev_str(s)
```

```python
def reverse(st):
  rev=""
  for i in range(len(st)-1,-1,-1):
    rev+=st[i]
  return rev
st=input()
print(reverse(st))
```

#example on default arguments

#example on keyword arguments

```python
def print_msg(name,branch):
  print("the name is ",name," and branch is ",branch);
print_msg(branch="AI&DS",name="chandu")
```

#example on variable length arguments(args):


```python
def add(*numbers):
  print(numbers)
  sum=0
  for i in numbers:
    sum+=i
  print(sum)
add(1,2,3,4,5)
add(1,2,3)
```

```python
#FUNCTIONS THAT RETURNS MULTIPLE VALUES
def multiple_values(*nums):
  s=sum(nums)
  c=len(nums)
  avg=s/c
  return s,c,avg
res_sum,res_count,res_avg=multiple_values(1,2,3,4)
print(f"sum is {res_sum},count is {res_count},the average is {res_avg}")
```
